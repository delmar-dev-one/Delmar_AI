```python
import pandas as pd
```

Dataframe de esportes

```python
sport = {
    'modalidades': ['futebol', 'basquete', 'vôlei', 'natação'],
    'quantidade de atletas': [11, 5, 6, 8],
    'nomes_dos_times': ['time A', 'time B', 'time C', 'time D']

}

# Criando um DataFrame a partir do dicionário com os dados do sport
df = pd.DataFrame(sport)

# Exibibindo 
print(df)
```

Transformando o conjunto de dados em csv com read


```python
df.to_csv('sport.csv', index=False) # sanvando em csv o dataframe de sport

#@ Somete para informar que o arquivo CSV foi criado com sucesso
print ('Arquivo CSV criado com sucesso!')
# Lendo o arquivo CSV e exibindo seu conteúdo
df_csv = pd.read_csv('sport.csv')
print(df_csv)
```

Convertendo em excel

```python
df.to_excel('sport.xlsx', index=False) # sanvando em excel o dataframe de sport

df_excel = pd.read_excel('sport.xlsx') # Lendo o arquivo Excel e exibindo seu conteúdo
df_excel
```

Criando outro datasets para concatenar

```python
jogadores = {
    'nomes_famosos': ['Pelé', 'Michael Jordan', 'Serena Williams', 'Usain Bolt'],
    'pais_de_origem': ['Brasil', 'Estados Unidos', 'Estados Unidos', 'Jamaica'],
}

df_jog = pd.DataFrame(jogadores)
df_jog
```

Concatenando

```python
df_atualizado = pd.concat([df, df_jog], ignore_index=True)


# preencher os valores ausentes com 'Desconecido'
df_atualizado.fillna('falta preencher', inplace=True)
print(df_atualizado)
```

json


```python
#===========================================================================================

df_json = df_atualizado.to_json(orient='records') # Convertendo o DataFrame para JSON
df_json

```

Fatiação de dados para se trabalha

```python
# Fatiamento de dados
#==========================================================================================
df_atualizado['modalidades'] # selecionando a coluna 'modalidades'
#==========================================================================================
df_atualizado.loc[1] # selecionando a segunda linha do DataFrame
#==========================================================================================
df_atualizado.loc[1:3] # selecionando as segundas 3 linhas do DataFrame
#==========================================================================================
df_atualizado.loc[1:5, 'modalidades'] # selecionando as segundas 5 linhas da coluna 'modalidades'
```

```python
df_filtando = df_atualizado['quantidade de atletas'] == "falta preencher" # criando um filtro para selecionar as linhas onde a quantidade de atletas é igual a "falta preencher"
df_filtrado = df_atualizado[df_filtando] # aplicando o filtro ao DataFrame
df_filtrado 
```

```python
filtro2 = (df_atualizado['nomes_famosos'] == "Pelé") & (df_atualizado['pais_de_origem'] == "Brasil") # criando um filtro para selecionar as linhas onde os nomes famosos são iguais a "Pelé" e o país de origem é igual a "Brasil"
filtro2 = df_atualizado[filtro2] # aplicando o filtro ao DataFrame
filtro2
```
